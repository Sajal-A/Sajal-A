```python
n = 6
for i in range(0,n):
    value = i
    for j in range(0,value+1): #number loop
        value=j+1
        print(value,end=' ') 
    for k in range(0,2*(n-i-1)): # space loop
        print(' ',end=' ')
    for l in range(0,value): # number loop
        print(value,end=' ')
        value -= 1
    
    print()
```

```python
#hollow square
nrows = 4
for i in range(1,nrows+1):
    for j in range(1,nrows+1):
        if (i==1 or i==nrows or j==1 or j==nrows):
            print('#',end=' ')
        else:
            print(' ',end=' ')
    print()
```

```python
no_of_rows = 5

for i in range(no_of_rows): # for number or rows
    for j in range(0,no_of_rows-i-1): #for number of spaces
        print(' ',end=' ')
    ch = 65
    for k in range(0,2*i+1): # for number of stars
        print(chr(ch),end=' ')
        if k < ((2*i+1)//2):
            ch+=1
        else:
            ch-=1
            
    for j in range(0,no_of_rows-i-1): #for number of spaces
        print(' ',end=' ')
    print()
```
