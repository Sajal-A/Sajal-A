4. Write a program to swap two variable using function

```python
def swap_variable(a, b) -> None:
  #using temporary variable
  # temp = a
  # a = b
  # b = temp
  #Swapping without temporary variable -- python miracle
  # a , b = b, a
  #using XOR operator
  a = a^b
  b = a^b
  a = a^b
  return a, b
num1 = 23
num2 = 45
print('Before swapping')
print(f'num1 : {num1} and num2 : {num2}')
num1, num2 = swap_variable(num1,num2)
print('After swapping')
print(f'num1 : {num1} and num2 : {num2}')
```

11. Write a program to print all prime number in certain interval {1-num (num = 1,2...)}

13. Write a program to print fibonnaci sequence

    0,1,1,2,3,5,8,13,34,55,89 ...

```python
nterms = int(input('How many terms? '))
n1 , n2 = 0, 1

counter = 0

if nterms <= 0:
  print('Please enter a valid term')
elif nterms == 1:
  print(n1)
else:
  print('Fibonacci Sequence: ')
  while counter<nterms:
    print(n1,end=' ')
    nth = n1+n2
    n1=n2
    n2=nth
    counter+=1
```

14. Write a program to check an armstrong number.

```python
number = '153'
length = len(number)
temp = int(number)
sum = 0
for i in range(length):
  rem = temp%10
  sum = sum + rem**length
  temp //= 10

if sum == int(number):
  print(f'{number} is an Armstrong number')
else:
  print(f'{number} is not an Armstrong number')
```

15. Write a program to find Armstrong in an interval

```python
lower = 10
upper = 10000
for num in range(lower, upper+1):
  power = len(str(num))
  temp = num
  sum = 0
  while temp > 0:
    rem = temp%10
    sum = sum + rem**power
    temp //= 10
  if sum == num:
    print(num,end=' ')
```

17. Write a program to find the HCF.

    HCF (highest common factor) or GCD (Greatest Common Division) is the smallest positive integer that divides two or more numbers without leaving a remainder.
    

```python
def compute_gcd(x,y):
  if x < y :
    small = x
  else:
    small = y
  for i in range(1, small+1):
    if (x%i == 0) and (y%i == 0):
      gcd = i
  return gcd

num1 = 54
num2 = 24
print(f'The HCF of {num1} and {num2} is {compute_gcd(num1,num2)}')
```

18. Write a program to find the LCM

  LCM or Least Common Multiple, is the smallest multiple that is exactly divible <br> by two or more numbers.
  

```python
def compute_lcm(x,y):
  if x > y:
    greater = x
  else:
    greater = y
  while True:
    if (greater%x == 0) and (greater%y == 0):
      lcm = greater
      break
    greater += 1
  return lcm

num1 = 54
num2 = 24
print(f'The LCM of {num1} and {num2} is {compute_lcm(num1,num2)}')
```
